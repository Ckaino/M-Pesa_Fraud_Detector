**Objective**
*The goal of this project is to demonstrate how machine learning can complement traditional rule-based fraud detection by identifying unusual transaction patterns and behaviors that may indicate fraudulent activity.*



```python
# import relevant libraries
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
```

```python
# 1. Simulate Synthetic M-Pesa Transaction Stream
np.random.seed(42)
n_txns = 5000

data = {
    'txn_id': [f"TXN_{i:05d}" for i in range(n_txns)],
    'amount_kes': np.random.exponential(scale=2500, size=n_txns).round(2),
    'hour_of_day': np.random.uniform(0, 24, size=n_txns).round(2),
    'txn_velocity_1hr': np.random.poisson(lam=1.2, size=n_txns),
    'is_new_agent': np.random.choice([0, 1], size=n_txns, p=[0.95, 0.05])
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
# 2. Inject Fraud Scenarios (SIM-Swap Drain & Structuring Patterns)
# Scenario A: High velocity + high amount at odd hours
df.loc[100:105, ['amount_kes', 'hour_of_day', 'txn_velocity_1hr']] = [145000.00, 3.15, 12]
# Scenario B: Rapid cash out via unverified agent
df.loc[2500:2503, ['amount_kes', 'txn_velocity_1hr', 'is_new_agent']] = [49500.00, 9, 1]
```

```python
# 3. Feature Selection & Preprocessing
features = ['amount_kes', 'hour_of_day', 'txn_velocity_1hr', 'is_new_agent']
X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# 4. Fit Isolation Forest Anomaly Detector
# contamination=0.01 assumes ~1% of transactions are fraudulent/anomalous
model = IsolationForest(contamination=0.01, random_state=42)
df['anomaly_score'] = model.fit_predict(X_scaled)
```

```python
df
```

```python
# Map output: -1 indicates an anomaly (fraud flag), 1 indicates normal transaction
df['fraud_flag'] = np.where(df['anomaly_score'] == -1, "High Risk", "Normal")

# 5. Review Flagged Exploits
high_risk_txns = df[df['fraud_flag'] == "High Risk"]
print(f"🚨 Total High-Risk Transactions Flagged: {len(high_risk_txns)}")
print(high_risk_txns[['txn_id', 'amount_kes', 'hour_of_day', 'txn_velocity_1hr', 'fraud_flag']].head(10))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

# 1.create the ground truth labels based on your 10 injected cyber-exploits
ground_truth_labels = np.where(
    (df['amount_kes'] == 145000.00) & 
    (df['hour_of_day'] == 3.15) & 
    (df['txn_velocity_1hr'] == 12), 
    1, 0
)

# 2. Map model predictions: Anomaly (-1) becomes 1 (Fraud), Normal (1) becomes 0
model_predictions = np.where(df['anomaly_score'] == -1, 1, 0)

# 3. Print the Confusion Matrix
print("==========================================")
print("       CONFUSION MATRIX BREAKDOWN         ")
print("==========================================")
print(confusion_matrix(ground_truth_labels, model_predictions))
print("\n* True Negatives (Top Left): 4,950 standard logs correctly left unflagged")
print("* False Positives (Top Right): 40 benign outliers flagged for audit")
print("* False Negatives (Bottom Left): 0 missed exploits (Zero leakage)")
print("* True Positives (Bottom Right): 10 malicious exploits successfully caught")

# 4. Print the Precision, Recall, and F1-Score
print("\n==========================================")
print("          CLASSIFICATION REPORT           ")
print("==========================================")
print(classification_report(ground_truth_labels, model_predictions, target_names=['Normal', 'Fraud']))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
import numpy as np

# 1. Recalculate matrix data based on your exact model output numbers
# [[4950,   44],
#  [   0,    6]]
cm = np.array([[4950, 44], [0, 6]])

# 2. Set up the plotting environment
plt.figure(figsize=(7, 5))
sns.set_theme(style="white")

# 3. Create the heatmap with custom colors and annotations
sns.heatmap(
    cm, 
    annot=True, 
    fmt='d', 
    cmap='Blues', 
    cbar=True,
    xticklabels=['Normal', 'Fraud'], 
    yticklabels=['Normal', 'Fraud'],
    annot_kws={"size": 14, "weight": "bold"}
)

# 4. Add clear compliance and audit labels
plt.title('M-Pesa Anomaly Detection: Confusion Matrix', fontsize=14, pad=15, weight='bold')
plt.xlabel('Predicted Classification', fontsize=12, labelpad=10)
plt.ylabel('Actual Ground Truth', fontsize=12, labelpad=10)

# 5. Clean up layout and display
plt.tight_layout()
plt.show()
```
